```python
# Convert Avg_Viewers into numeric format
df["Avg_Viewers"] = pd.to_numeric(df["Avg_Viewers"],errors="coerce")
# Check invalid negative values
print((df["Avg_Viewers"] < 0).sum())
```

```python
#Check whether HD_Available contains only Yes and No.
print(df["HD_Available"].value_counts())
print(df["HD_Available"].isin(["Yes", "No"]).all())
```

```python
#Identify NA, N/A, unknown, and blank values
print(df.isin(["NA","N/A","na","n/a","unknown","Unknown",""]).sum())
```

```python
#Identify columns where numeric values are stored as text
print(df.dtypes)

#specify numeric columns
num_cols = ["Avg_Viewers","Total_Likes","Social_Media_Followers",
            "YouTube_Views","Monthly_Reach"]


for col in num_cols:df[col] = pd.to_numeric(df[col],errors="coerce")

for col in num_cols:
    print(col, ":", df[col].dtype)
```

```python
#Standardize the Channel_Name values
df["Channel_Name"] = (df["Channel_Name"].str.strip().str.title().
                      str.replace("Tv", "TV"))
print(df["Channel_Name"])
```

```python
#Handling missing values
# Numerical columns → median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns → mode
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())
```

```python
# Fill ratings with median
df["Avg_Rating"] = df["Avg_Rating"].fillna(df["Avg_Rating"].median())
print("Avg Rating")
print(df["Avg_Rating"])

df["Subscription_Fee"] = df["Subscription_Fee"].fillna(df["Subscription_Fee"].median())
print("Subscription Fee")
print(df['Subscription_Fee'])

print("\nMInium Subscription Fee")
print(df["Subscription_Fee"].min())
```

```python
# Handling unrealistic Years_Active values.
df["Years_Active"] = df["Years_Active"].fillna(df["Years_Active"].median())

print(df["Years_Active"].min())
```

```python
# Fill numerical missing values
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

print(df[col])

# Fill categorical missing values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df[col])
```

```python
# Check missing values
print("After Cleaning Missing Values:")
print(df.isnull().sum())

# Check data types
print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv("cleaned_tv_channel_dataset.csv",index=False)
print("Cleaned dataset saved successfully")
```
